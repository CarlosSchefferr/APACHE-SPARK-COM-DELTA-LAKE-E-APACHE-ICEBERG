# Delta Lake no Apache Spark

Notebook demonstrando CREATE, INSERT, UPDATE e DELETE em tabela Delta.


```python
from pyspark.sql import SparkSession
from delta import configure_spark_with_delta_pip

builder = (
    SparkSession.builder
    .appName('delta-lake-demo')
    .master('local[*]')
    .config('spark.sql.extensions', 'io.delta.sql.DeltaSparkSessionExtension')
    .config('spark.sql.catalog.spark_catalog', 'org.apache.spark.sql.delta.catalog.DeltaCatalog')
)

spark = configure_spark_with_delta_pip(builder).getOrCreate()
```

```python
spark.sql("CREATE DATABASE IF NOT EXISTS demo")
spark.sql("USE demo")

spark.sql('''
CREATE TABLE IF NOT EXISTS pedidos_delta (
  pedido_id STRING,
  cliente_id STRING,
  status STRING,
  valor_total DOUBLE,
  data_pedido DATE
) USING delta
LOCATION './warehouse/delta/pedidos_delta'
''')
```

```python
# INSERT
spark.sql('''
INSERT INTO pedidos_delta VALUES
('P001','C001','criado',120.50,DATE '2026-01-10'),
('P002','C002','criado',89.90,DATE '2026-01-11')
''')

# UPDATE
spark.sql("UPDATE pedidos_delta SET status='faturado' WHERE pedido_id='P001'")

# DELETE
spark.sql("DELETE FROM pedidos_delta WHERE pedido_id='P002'")

spark.sql("SELECT * FROM pedidos_delta").show(truncate=False)
```
